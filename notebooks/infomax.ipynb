```python
from mne import create_info, io,preprocessing
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore
```

```python
n = 651
channels = [str(int) for int in np.arange(n)]
print(channels)
```

```python
f_s = 5.3
```

```python
traces = zscore(np.load('fluo_220127_F4_run2.npy'))
```

```python
info = create_info(channels,f_s)
```

```python
raw = io.RawArray(traces, info)
```

```python
%matplotlib notebook
raw.plot()
```

```python
ica = preprocessing.ICA(651,method='infomax')
```

```python
ica.fit(raw)
```
